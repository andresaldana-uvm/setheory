Consideremos el conjunto $U$ formado por los estudiantes de una universidad. Este conjunto será nuestro universo de referencia.

A partir de $U$, podemos definir subconjuntos usando propiedades que determinan qué estudiantes pertenecen a cada conjunto. En la librería `setheory`, esta idea se representa mediante un objeto `Set`, que recibe un universo y un predicado.

$$
U = \set{\text{Ana}, \text{Bruno}, \text{Carla}, \text{Diego}, \text{Elena}, \text{Fernando}}.
$$

```python
from setheory.api import Set
```

```python
U = {"Ana", "Bruno", "Carla", "Diego", "Elena", "Fernando"}
```

Supongamos que queremos identificar a los estudiantes que pertenecen al grupo de estudiantes de matemáticas. Sea $M$ el conjunto formado por aquellos estudiantes que cumplen esta condición.

Podemos escribir este conjunto mediante un predicado $p$, donde $p$ es verdadero exactamente cuando $x$ pertenece al grupo de matemáticas.

$$
M = \set{x \in U \vert p}
$$

```python
M = Set(U, lambda x: x in {"Ana", "Carla", "Fernando"})

M
```

Una vez definido $M$, podemos preguntar si un estudiante pertenece al conjunto. Para que $x$ pertenezca a $M$, deben cumplirse dos condiciones: $x$ debe pertenecer al universo $U$ y debe satisfacer el predicado que define a $M$.

$$
x \in M \iff x \in U \land p.
$$

```python
"Ana" in M, "Bruno" in M
```

```python
# También podemos comprobar qué ocurre con un elemento que ni siquiera pertenece al universo
"Gabriel" in M

# Aquí el resultado debe ser `False`, independientemente de que el predicado pudiera aceptar
# ese valor.
```

Aunque $M$ está definido mediante un predicado, en un universo finito podemos obtener explícitamente los elementos que lo forman. La propiedad `elements` recorre el universo y conserva a aquellos elementos para los que el predicado es verdadero.

$$
\text{elements}_M = \set{x \in U \vert p}.
$$

```python
M.elements
```

Consideremos ahora a los estudiantes que participan en actividades deportivas. Sea $S$ el subconjunto de $U$ definido por esta propiedad.

Al igual que antes, el conjunto no se construye proporcionando directamente sus elementos a `Set`. En su lugar, proporcionamos una función que determina si cada elemento del universo pertenece al conjunto.

$$
S = \set{x \in U \vert q}.
$$

```python
S = Set(U, lambda x: x in {"Bruno", "Carla", "Elena"})

S.elements
```

Los conjuntos definidos sobre un mismo universo no tienen por qué ser disjuntos. Un mismo estudiante puede satisfacer las condiciones que definen varios conjuntos.

En nuestro ejemplo, Carla pertenece tanto al conjunto de estudiantes de matemáticas como al conjunto de estudiantes que participan en actividades deportivas.

$$
\text{Carla} \in M, \qquad \text{Carla} \in S.
$$

```python
"Carla" in M, "Carla" in S
```

También podemos definir un conjunto que no contenga ningún elemento. Para ello, basta con utilizar un predicado que nunca sea verdadero.

El conjunto obtenido es el conjunto vacío, que detonamos por $\emptyset$.

$$
\emptyset = \set{x \in U \vert \perp}.
$$

```python
emptyset = Set(U, lambda x: False)

emptyset.elements
```

```python
# Podemos comprobar que ningún estudiante pertenece al conjunto:
"Ana" in emptyset
```

En el extremo opuesto, podemos definir un conjunto cuyos elementos sean todos los elementos del mismo universo. Para ello utilizaremos un predicado que siempre sea verdadero.

Este conjunto coincide con el universo $U$.

$$
U = \set{x \in U \vert \top}.
$$

```python
allset = Set(U, lambda x: True)

allset.elements
```

```python
# Podemos comprobar que todos los estudiantes pertenecen a este conjunto:
"Ana" in allset
```

La forma en la que escribimos un predicado no determina por sí sola la identidad del conjunto. Lo importante son los elementos que finalmente pertenecen a él.

Por ejemplo, podemos definir el conjunto de estudiantes de matemáticas utilizando directamente sus nombres, o mediante otro predicado que produzca exactamente los mismos elementos.

$$
A = \set{x \in U \vert p}, \qquad B = \set{x \in U \vert q}.
$$

Si

$$
\forall{x} \in U, \qquad p \iff q.
$$

```python
set_by_name = Set(U, lambda x: x in {"Ana", "Carla", "Fernando"})
set_by_condiction = Set(U, lambda x: x in M)

set_by_name == set_by_condiction
```
