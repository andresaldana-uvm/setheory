Sea $U$ el conjunto de estudiantes de una universidad. Consideremos dos subconjuntos de $U$, definidos a partir de propiedades que pueden cumplir los estudiantes.

En particular, sean $A$ y $B$ los conjuntos formados por los estudiantes que cumplen dos condiciones distintas. A partir de estos conjuntos podemos estudiar las principales operaciones de la teoría de conjuntos y ver cómo se representan mediante `Set`.

```python
from setheory.api import Set
```

```python
U = {"Ana", "Bruno", "Carla", "Diego", "Elena", "Fernando"}

A = Set(U, lambda x: x in {"Ana", "Carla", "Elena"})
B = Set(U, lambda x: x in {"Carla", "Diego", "Elena"})

A, B
```

La intersección de $A$ y $B$ contiene los elementos que pertenecen a ambos conjuntos. Por lo tanto, un elemento pertenece a la intersección cuando cumple al mismo tiempo las condiciones que definen a $A$ y a $B$.

$$
A \cap B = \set{x \in U \vert p \land q}.
$$

```python
A & B
```

La unión de $A$ y $B$ contiene los elementos que pertenecen a al menos uno de los dos conjuntos. Esto incluye tanto los elementos que aparecen únicamente en uno de ellos como los que pertenecen a ambos.

$$
A \cup B = \set{x \in U \vert p \lor q}.
$$

```python
A | B
```

La diferencia entre $A$ y $B$ contiene los elementos que pertenecen a $A$, pero no a $B$. A diferencia de la unión y la intersección, esta operación depende del orden de los conjuntos.

Por ejemplo, $A - B$ y $B - A$ contienen, en general, elementos diferentes.

$$
A - B \set{x \in U \vert p \land \neg{q}}.
$$

```python
A - B, B - A
```

La diferencia simétrica entre $A$ y $B$ contiene los elementos que permanecen exactamente a uno de los dos conjuntos. Por ello, los elementos que pertenecen a ambos conjuntos quedan fuera del resultado.

También podemos obtenerla uniendo las dos diferencias anteriores.

$$
A \Delta B = \set{x \in U \vert p \neq q}.
$$

Equivalentemente,

$$
A \Delta B = (A - B) \cup (B - A).
$$

```python
A ^ B
```

```python
(A ^ B) == (A - B) | (B - A)
```

Para definir el complemento necesitamos fijar un universo. El complemento de $A$ contiene todos los elementos de $U$ que no pertenecen a $A$.

En este caso, el complemento representa a los estudiantes del universo que no cumplen la condición que define a $A$.

$$
A^c = \set{x \in U \vert \neg{p}}.
$$

```python
~A
```

Algunas operaciones pueden expresarse utilizando otras operaciones de conjuntos. Por ejemplo, para obtener $A - B$, podemos tomar los elementos de $A$ y eliminar los elementos que pertenecen a $B$. Esto es equivalente a tomar $A$ y quedarse únicamente con los elementos que no pertenecen a $B$.

Por tanto, la diferencia puede escribirse como una intersección con el complemento de $B$:

$$
A - B = A \cap B^c.
$$

```python
(A - B) == (A & ~B)
```

De forma similar, la diferencia simétrica puede obtenerse tomando la unión de $A$ y $B$ y eliminando los elementos que pertenecen a ambos:

$$
A \Delta B = (A \cup B) - (A \cap B).
$$

```python
(A ^ B) == (A | B) - (A & B)
```
