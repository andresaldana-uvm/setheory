Continuaremos trabajando con el conjunto $U$ de estudiantes de una universidad. A partir de este universo consideremos tres conjuntos por definidos por distintas condiciones académicas.

Sea $A$ el conjunto de estudiantes que cursan matemáticas, $B$ el conjunto de estudiantes que cursan matemáticas o física, y $C$ el conjunto de estudiantes que cursan física.

Estos conjuntos nos permitirán estudiar cuándo un conjunto está contenido en otro y cuándo dos conjuntos tienen exactamente los mismos elementos.

$$
\begin{align}
A &= \set{x \in U \vert p} \\
B &= \set{x \in U \vert q} \\
C &= \set{x \in U \vert r}
\end{align}
$$


```python
from setheory.api import Set
```

```python
U = {"Ana", "Bruno", "Carla", "Diego", "Elena", "Fernando"}

p = {"Ana", "Carla"}
q = {"Carla", "Diego", "Elena"}

A = Set(U, lambda x: x in p)
B = Set(U, lambda x: x in p | q)
C = Set(U, lambda x: x in q)

A.elements, B.elements, C.elements
```

Decimos que $A$ es un subconjunto de $B$ si todos los elementos de $A$ también pertenecen a $B$. Esta relación se escribe $A \subseteq B$.

En términos de los estudiantes, esto significa que cualquier estudiante que pertenece a $A$ también pertenece a $B$.

Formalmente,

$$
A \subseteq B = \iff \forall{x} \in U, \qquad x \in A \Rightarrow x \in B.
$$

```python
A <= B
```

```python
# Podemos interpretar el resultado directamente:
print(A.elements)
print(B.elements)

# Como todos lo estudiantes de `A` también aparecen en `B`, la relación A ⊆ B
# es verdadera.
```

La relación anterior no es simétrica. Que $A \subseteq B$ no implica que $B \subseteq A$.

Para que $B$ sea subconjunto de $A$, cada elemento de $B$ tendría que pertenecer también a $A$. Basta encontrar un elemento $B$ que no pertenezca a $A$ para que la relación falle.

$$
B \nsubseteq A.
$$

```python
B <= A
```

```python
# Podemos obserar los elementos responsables de esta diferencia:
B.elements - A.elements
```

Un conjunto $A$ es un subconjunto propio de $B$ cuando todos sus elementos pertenecen a $B$, pero $A$ y $B$ no contienen exactamente los mismos elementos.

Esta relación se escribe $A \subset B$ y puede expresarse como:

$$
A \subset B \iff A \subseteq B \land A \neq B.
$$

```python
A < B
```

Todo conjunto es subconjunto de sí mismo, ya que cada elemento del conjunto pertenece evidentemetne al propio conjunto.

Sin embargo, ningún conjunto es subconjunto propio de sí mismo. La diferencia está precisamente en la exigencia de que los dos conjuntos no sean iguales.

$$
A \subseteq A,
$$

pero

$$
A \not\subset A.
$$

```python
print(A <= A)
print(A < A)
```

Dos conjuntos son iguales cuando contienen exactamente los mismos elementos. No es necesario que hayan sido construidos utilizando el mismo predicado; basta con que definan la misma colección de elementos.

Formalmente,

$$
A = B \iff \forall{x} \in U, \qquad (x \in A \iff x \in B).
$$

Para verlo, podemos definir nuevamente el conjunto de estudiantes de matemáticas utilizando un predicado diferente.

```python
D = Set(U, lambda x: x in {"Ana", "Carla"})

D.elements
```

```python
A == D
```

La igualdad de conjuntos también puede caracterizarse mediante la inclusión en ambas direcciones. Dos conjuntos son iguales exactamente cuando cada uno es un subconjunto del otro.

Por tanto,

$$
A = B \iff (A \subseteq B) \land (B \subseteq A).
$$

```python
(A <= D) and (D <= A)
```

No todos los pares de conjuntos definidos sobre un mismo universo son comparables mediante la relación de inclusión.

Consideremos ahora los conjuntos $A$ y $C$, correspondientes a estudiantes de matemática y física, respecivamente. Carla pertenece a ambos conjuntos, pero cada uno contiene además elementos que no pertenecen al otro.

En consecuencia, ninguno de los dos conjuntos está contenido en el otro.

$$
A \nsubseteq C \qquad \text{y} \qquad C \nsubseteq A
$$

```python
print(A <= C)
print(C <= A)
```

```python
# Podemos observar directamente los elementos que impiden cada inclusión:
print(A.elements - C.elements)
print(C.elements - A.elements)
```

Dos casos aparecen de forma natural en la relación de inclusión. El conjunto vacío es subconjunto de cualquier conjunto, ya que no contiene ningún elemento que pueda incumplir la condición de inclusión.

Por otro lado, todo subconjunto de $U$ está contenido en $U$.

$$
\emptyset \subseteq A, \qquad A \subseteq U.
$$

```python
emptyset = Set(U, lambda x: False)
universe_set = Set(U, lambda x: True)

print(emptyset <= A)
print(A <= universe_set)
```

Las relaciones estudiadas pueden resumirse a partir de la cantidad de elementos que comparten dos conjuntos. Si todos los elementos de $A$ pertenecen a $B$, tenemos $A \subseteq B$. Si además existe algún elemento de $B$ que no pertenece a $A$, entonces $A \subset B$. Finalmente, si la inclusión se cumple en ambas direcciones, los conjuntos son iguales.

De esta manera, las tres relaciones están estrechamente relacionadas:

$$
A = B \Longrightarrow A \subseteq B \quad \text{y} \quad B \subseteq A,
$$

mientras que

$$
A \subset B \Longrightarrow A \subseteq B \quad \text{y} \quad A \neq B.
$$

```python
print(A <= B)
print(A < B)
print(A == D)
```
