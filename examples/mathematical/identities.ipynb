Sea $U$ el conjunto universal de estudiantes de una universidad. Para cada estudiante $x \in U$, consideremos distintas propiedades que permiten definir subconjuntos de $U$.

Utilizaremos estos subconjuntos para verificar experimentalmente algunas identidades fundamentales de la teoría de conjuntos. Cada identidad se expresará primero en términos matemáticos y posteriormente comprobará utilizando la implementación de `Set`.

La idea central es que dos expresiones de conjuntos son iguales cuando contienen exactamente los mismos elementos:

$$
A = B \iff \forall{x} \in U, \qquad x \in A \iff x \in B
$$

```python
from setheory.api import Set
```

Sea

$$
U = \set{\text{Ana}, \text{Bruno}, \text{Carla}, \text{Diego}, \text{Elena}, \text{Fernado}}
$$

el conjunto de estudiantes considerado.

Definimos los subconjuntos $A$, $B$ y $C$ mediante propiedades distintas de los estudiantes. En particular, sea $A$ el conjunto de estudiantes de ingeniería, $B$ el conjunto de estudiantes que participan en actividades extracurriculares y $C$ el conjunto de estudiantes que han aprobado todas sus asignaturas.

Así,

$$
A = \set{x \in U \vert p}, \qquad B = \set{x \in U \vert q}, \qquad C = \set{x \in U \vert r}.
$$

```python
U = {"Ana", "Bruno", "Carla", "Diego", "Elena", "Fernando"}

p = {"Ana", "Bruno", "Diego"}
q = {"Ana", "Carla", "Diego", "Fernando"}
r = {"Ana", "Bruno", "Carla"}

A = Set(U, lambda x: x in p)
B = Set(U, lambda x: x in q)
C = Set(U, lambda x: x in r)

A.elements, B.elements, C.elements
```

Consideremos primero la unión de un conjunto consigo mismo. Para cualquier $A \subseteq U$, la teoría de conjuntos establece que

$$
A \cup A = A.
$$

En términos de pertenencia, para todo $x \in U$,

$$
x \in A \cup A \iff (a \in A) \lor (x \in A) \iff x \in A.
$$

Por tanto, aplicar la unión dos veces no incorpora ningún elemento nuevo.

```python
A_union = A | A

print(A.elements)
print(A_union.elements)
print(A_union == A)
```

Las operaciones de unión e intersección está relacionadas mediante las leyes de absorción. Una de ellas establece que

$$
A \cup (A \cap B) = A.
$$

Intuitivamente, $A \cap B$ solo contiene estudiantes que ya pertenecen a $A$. Por ello, al unir $A \cap B$ con $A$, no aparece ningún estudiante que no estuviera previamente en $A$.

Formalmente,

$$
x \in A \cup (A \cap B) \iff (x \in A) \lor ((x \in A) \land (x \in B)),
$$

y por la ley lógica de absorción,

$$
(p \lor (p \land q)) \iff p.
$$

```python
absorption = A | (A & B)

print(A.elements)
print(absorption.elements)
print(absorption == A)
```

La intersección distribuye sobre la unión. Para $A, B, C \subseteq U$,

$$
A \cap (B \cup C) = (A \cap B) \cup (A \cap C).
$$

Para un estudiante $x$, pertenecer al lado izquierdo significa que $x$ pertenece a $A$ y, además, pertenece a $B$ o a $C$:

$$
x \in A \cap (B \cup C) \iff p \land (q \lor r).
$$

La expresión lógica equivalente es

$$
p \land (q \lor r) \iff (p \land q) \lor (p \land r),
$$

que corresponde exactamente al lado derecho de la identidad.

```python
left = A & (B | C)
right = (A & B) | (A & C)

print(left.elements)
print(right.elements)
print(left == right)
```

Las leyes de De Morgan relacionan el complemento con la unión y la intersección. Consideremos

$$
(A \cup B)^c = A^c \cap B^c.
$$

El lado izquierdo contiene a los estudiantes que no pertenecen ni a $A$ ni a $B$. Por otro lado, el lado derecho exigue simultáneamente que el estudiante no pertenezca a $A$ y que tampoco pertenezca a $B$.

Para cada $x \in U$,

$$
x \in (A \cup B)^c \iff \neg{(p \lor q)}
$$

Por la ley de De Morgen de la lógica proposicional,

$$
\neg{p \lor q} \iff \neg{p} \land \neg{q},
$$

y, por tanto,

$$
(A \cup B)^c = A^c \cap B^c.
$$

```python
left = ~(A | B)
right = (~A) & (~B)

print(left.elements)
print(right.elements)
print(left == right)
```

De manera análoga, el complemento de una intersección puede expresarse mediante la unión de los complementos:

$$
(A \cap B)^c = A^c \cup B^c.
$$

Para $x \in U$,

$$
x \in (A \cap B)^c \iff \neg{(p \land q)}. 
$$

Nuevamente, por una ley de De Morgan de la lógica,

$$
\neg{(p \land q)} \iff \neg{p} \lor \neg{q}.
$$

Por ello,

$$
(A \cap B)^c = A^c \cup B^c.
$$

```python
left = ~(A & B)
right = (~A) | (~B)

print(left.elements)
print(right.elements)
print(left == right)
```

El complemento se define respecto al universo U. Por ello, complementar dos veces un conjunto recupera el conjunto original:

$$
(A^c)^c = A.
$$

En términos de predicados,

$$
x \in (A^c)^c \iff \neg{(\neg{p})} \iff p
$$

La identidad muestra que el complemento actúa como una operación involutiva.

```python
double_complement = ~(~A)

print(A.elements)
print(double_complement.elements)
print(double_complement == A)
```

La diferencia de conjuntos puede expresarse utilizando únicamente intersección y complemento. Para $A, B \subseteq U$,

$$
A - B = A \cap B^c.
$$

En efecto, un estudiante pertenece a $A - B$ exactamente cuando pertenece a $A$ y no pertenece a $B$:

$$
x \in A - B \iff p \land \neg{q}.
$$

Pero $x \in B^c$ si solo si $\neg{q}$. Por tanto,

$$
p \land \neg{q} \iff x \in A \cap B^c.
$$

```python
left = A - B
right = A & (~B)

print(left.elements)
print(right.elements)
print(left == right)
```

La diferencia simétrica $A \Delta B$ contiene los elementos que pertenecen exactamente a uno de los dos conjuntos.

Puede expresarse mediante dos diferencias:

$$
A \Delta B = (A - B) \cup (B - A).
$$

La condición de pertenencia es

$$
x \in A \Delta B \iff (x \in A \land x \notin B) \lor (x \in B \land x \notin A).
$$

Es decir, $x$ pertenece a la diferencia simétrica cuando pertenece a $A$ o a $B$, pero no a ambos.

```python
left = A ^ B
right = (A - B) | (B - A)

print(left.elements)
print(right.elements)
print(left == right)
```

Finalmente, consideremos el universo $U$ como el conjunto que contiene todos los elementos posibles del dominio. La unión de cualqueir subconjunto $A \subseteq U$ con $U$ es el propio universo:

$$
A \cup U = U.
$$

De forma dual,

$$
A \cap U = A.
$$

Estas identidades expresan que $U$ actúa como elemento neutro de la intersección y como elemento absorbente de la unión.

```python
U = Set(U, lambda x: True)

print((A | U).elements)
print((A & U).elements)

print((A | U) == U)
print((A & U) == A)
```
