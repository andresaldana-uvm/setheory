```python
# tools/notebooks/concept.ipynb
from collections.abc import Callable
```

```python
class Set:
    def __init__(self, U, P: Callable):
        self.U = U
        self.P = P

    def __contains__(self, x):
        return x in self.U and self.P(x)

    @property
    def elements(self) -> set:
        return {x for x in self.U if self.P(x)}

    # set operations
    def __and__(self, B):
        self._validate(B)
        return Set(self.U, lambda x: self.P(x) and B.P(x))

    def __or__(self, B):
        self._validate(B)
        return Set(self.U, lambda x: self.P(x) or B.P(x))

    def __sub__(self, B):
        self._validate(B)
        return Set(self.U, lambda x: self.P(x) and not B.P(x))

    def __xor__(self, B):
        self._validate(B)
        return Set(self.U, lambda x: self.P(x) != B.P(x))

    def __invert__(self):
        return Set(self.U, lambda x: not self.P(x))

    # set-relations
    def __le__(self, B):
        self._validate(B)
        return self.elements <= B.elements

    def __lt__(self, B):
        self._validate(B)
        return self.elements < B.elements

    def __eq__(self, B):
        if not isinstance(B, Set):
            return NotImplemented
        self._validate(B)
        return self.elements == B.elements

    # private methods
    def _validate(self, B):
        if self.U is not B.U:
            raise ValueError("Sets must be the same universe.")

    def __repr__(self):
        return f"Set({self.elements!r})"

    def __str__(self):
        return str(self.elements)
```

```python
U = set(range(1, 11))

A = Set(U, lambda x: x % 2 == 0)
B = Set(U, lambda x: x > 5)
C = Set(U, lambda x: x % 3 == 0)
```

```python
print(4 in A)
print(5 in A)
print(6 in B)
print(3 in C)
print(11 in A)
```
